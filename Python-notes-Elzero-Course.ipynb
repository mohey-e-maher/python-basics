### String Methods

```python
# len() => show the length of the string
# -----------------------------------
print(len(myString))

# strip() , rstrip() , lstrip()
# -----------------------------------
stripo = "     momo    "
print(stripo.strip())
print(stripo.rstrip())
print(stripo.lstrip())

stripcharacter = "#####momo#####"
print(stripcharacter.strip("#"))
print(stripcharacter.rstrip("#"))
print(stripcharacter.lstrip("#"))

# title() used to make the first character Capitalized
# -----------------------------------
update = "momo is noT a Slave"
print(update.title())

# capitalized() make all characters capital 
# -----------------------------------
update2 = "momo is noT a Slave!!!!!!!!!"
print(update2.capitalize())

# zfill() it's a function used to make the digits all in the same weight 
# -----------------------------------
a,b,c = "1","11","111"
print(a.zfill(4))
print(b.zfill(4))
print(c.zfill(4))
```

```python
# Split() method 
# -----------------------------------
a = "I Love Python and PHP"
print(a.split())
# back with list of the words in string and the default seperator is space " "
b = "I-Love-Python-and-PHP"

# make spilit for only 3 elements from the left
print(b.split("-",3))
# make spilit for only 3 elements from the left
print(b.rsplit("-",2))

# center() Method
# -----------------------------------
E = "Mohey"
print(E.center(11,"+"))

# count() Method => count the number of character exists
# -----------------------------------
name = "Mohey Eldeen Maher Mohey Eldeen"
print(name.count("Mohey Eldeen"))

# swapcase() Method => it makes all upper to lower and reverse
# -----------------------------------
print("HeLLo MEEmo".swapcase())

# startswith() Method => check if the string start with this string ?
# -----------------------------------
mega = "Hello My Brother"
print(mega.startswith("Hello"))
print(mega.startswith("M",6,13))

# endswith() Method => check if the string start with this string ?
# -----------------------------------
mega = "Hello My Brother"
print(mega.endswith("Brother"))
print(mega.endswith("r",6,30))
```

```python
# index() Method => Back with the index of the 
# -----------------------------------
a = "I Love Python"
print(a.index("P")) # index of P
# print(a.index("P",0,5)) # Error without handling
print(a.index("P",0,10)) # index of P

# find() Method => Back with -1 if the substring not exists
# -----------------------------------
b = "I Love Python"
print(b.find("P")) # index of P
print(b.find("P",0,5))

# rjust(width, Fill character),ljust() are used to justify the string right or left 
# -----------------------------------
print(b.rjust(20,">"))
print(b.ljust(20,"<"))

# spilitlines() Method => back with list of string lines
# -----------------------------------
E = '''First Line
Second Line
Third Line'''
E = "Hello\nmy\nbrother"
print(E.splitlines())

# expandtabs() Method => used to change the number of spaces in tabs of strings
# -----------------------------------
gg= "Hello\tmy\texpensive"
print(gg.expandtabs(20))

# check Functions
one = "I Love Python And 2D"
print(one.istitle())

two = "  "
print(two.isspace())

Three = "I love Python"
print(Three.islower())

x= "AaaaBbbbb"
print(x.isalpha())
```

### List Methods

```python
# Append() Function
# -----------------------------
myLisst.append("Hello")
print(myLisst)
myLisst.append(["Hello","my","Friend"])
print(myLisst[7][2])

# extend() Function >>  make all list as one serial 
# -----------------------------
myLisst.extend(["A","B","C"])

# sort() Function
# -----------------------------
y = [1,2,100,120,-10,17,29]
print(y)
y.sort()
print(y)
y.sort(reverse=True)
print(y)

# reverse() Function
# -----------------------------
z=[10,1,9,80,100,"Mohey"]
z.reverse()
print(z)

# Clear() Function >> make the list empty
# -----------------------------
z.clear()
print(z)

# Copy() Function >> make copy of the list in another one
# -----------------------------
z= y.copy()
print(z)
```

```python
# count() Function >> count how many time element exists
# -----------------------------
print(myLisst.count("A"))

# index() Function >> get the index of element
# -----------------------------
print(z.index(29))

# insert() Function >> add value before the element
# -----------------------------
f =[1,2,3,4,5,"A","B"]
f.insert(1,1.5)
print(f)

# pop() Function >> remove the item from the list and return it
# -----------------------------
g=[1,2,3,4,5,6,7]
g.pop()
print(g)
```

### Set Methods

```python
a = {1,2,3}
a.clear()
print(a)

# union >> To make 2 sets in one set
a = {1,2,3}
b = {4,5,6,7}
d ={"mega","Mohey"}
print(a | b)
c = a.union(b,d)
print(c)
```

```python
# [1] copy() Method
m= c.copy()
print(m)

# [2] add() Method
m.add(88)
print(m)

# [3] remove() Method
m.remove("mega")
print(m)
# m.remove("mega")

#############if Element not Exist will get Error##############
# to skip this error we use >>>> discard()
m.discard("mega")
print(m)

# [4] pop() >> remove element in random way
z = m.pop()
print(z)
```

```python
# [5] Diffrence() 
a = {1,2,3,4}
b= {1,2,3,"Mega","mohey"}
print(a)
print(a.difference(b)) # a-b
print(a)
print(40*"=")

# [6] Diffrence_update() 
m = {1,2,3,4}
n= {1,2,3,"Mega","mohey"}
print(m)
print(m.difference_update(n)) # a-b
print(m)
print(40*"=")

# [7] intersection()
e = {1,2,3,4,"x"}
f= {"Osama","x",2}
print(e.intersection(f))
print(40*"=")

# [8] intersection_update()
e = {1,2,3,4,"x"}
f= {"Osama","x",2}
print(e)
e.intersection_update(f)
print(e)
print(40*"=")

# [9] Symmetric_diffrence()
i={1,2,3,4,5,"X"}
j = {"Osama","Zero",1,2,4,"X"}
print(i)
print(i.symmetric_difference(j))
print(i)
```
